```python
import pandas as pd
import os

# 파일 경로 설정 (노트북 위치 기준: 03_배포용/01_train/00_preprocess/)
# 데이터 폴더: 03_배포용/00_data/00_raw_data/
base_path = '../../00_data/00_raw_data/'
df_car_oct_path = os.path.join(base_path, 'df_car_oct.csv')
df_car_nov_path = os.path.join(base_path, 'df_car_nov.csv')
output_path = os.path.join(base_path, '../01_combined_data/df_car_combined.csv')

# CSV 파일 읽기
print("파일 읽는 중...")
df_car_oct = pd.read_csv(df_car_oct_path, encoding='utf-8')
df_car_nov = pd.read_csv(df_car_nov_path, encoding='utf-8')

print(f"df_car_oct.shape: {df_car_oct.shape}")
print(f"df_car_nov.shape: {df_car_nov.shape}")
print(f"\ndf_car_oct 컬럼: {list(df_car_oct.columns)}")
print(f"\ndf_car_nov 컬럼: {list(df_car_nov.columns)}")
```

```python
# 지정된 컬럼 순서로 정렬
desired_column_order = [
    '등록번호', '차량번호', '제조사', '대표차종명', '모델', '등급', '세부등급', 
    '제조사코드', '모델코드', '등급코드', '세부등급코드', '색상', '연월식', 
    '신차가격(선택옵션포함)', '배기량', '차종', '연료', '주행거리', 
    '최초 광고 등록일', '최초 광고가', '판매신고일', '판매신고가', 
    '용도변경이력', '사고유무', '단순수리', '교환부위', '판금부위', 
    '선택 추가 옵션1', '선택 추가 옵션2', '선택 추가 옵션3', '선택 추가 옵션4', 
    '선택 추가 옵션5', '선택 추가 옵션6', '선택 추가 옵션7', '선택 추가 옵션8', 
    '선택 추가 옵션9', '선택 추가 옵션10', '선택 추가 옵션11', '선택 추가 옵션12', 
    '선택 추가 옵션13'
]

# 공통 컬럼 확인
common_columns = set(df_car_oct.columns) & set(df_car_nov.columns)
print(f"공통 컬럼 개수: {len(common_columns)}")

# 지정된 순서 중 존재하는 컬럼만 필터링
ordered_columns = [col for col in desired_column_order if col in common_columns]

# 누락된 컬럼이 있는지 확인
missing_cols = common_columns - set(ordered_columns)
if missing_cols:
    print(f"\n경고: 다음 컬럼들이 지정된 순서에 없어 추가됩니다: {missing_cols}")
    ordered_columns.extend(sorted(missing_cols))

print(f"\n최종 컬럼 개수: {len(ordered_columns)}")
print(f"컬럼 순서: {ordered_columns[:5]}...{ordered_columns[-5:]}")

# 컬럼 순서를 지정된 순서로 맞춤
df_car_oct_aligned = df_car_oct[ordered_columns]
df_car_nov_aligned = df_car_nov[ordered_columns]

print(f"\n정렬 후 df_car_oct.shape: {df_car_oct_aligned.shape}")
print(f"정렬 후 df_car_nov.shape: {df_car_nov_aligned.shape}")
```

```python
# 행 기준으로 합치기 (vertical concatenation)
df_combined = pd.concat([df_car_oct_aligned, df_car_nov_aligned], axis=0, ignore_index=True)

print(f"합쳐진 데이터프레임 shape: {df_combined.shape}")
print(f"\n첫 5행:")
print(df_combined.head())
print(f"\n마지막 5행:")
print(df_combined.tail())
```

```python
# 합쳐진 데이터 저장
df_combined.to_csv(output_path, index=False, encoding='utf-8')
print(f"저장 완료: {output_path}")
print(f"총 행 수: {len(df_combined)}")
print(f"총 컬럼 수: {len(df_combined.columns)}")
```
